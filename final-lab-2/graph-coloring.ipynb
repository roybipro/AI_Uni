```python
# Graph (Australia map)
graph = {
    'WA': ['NT', 'SA'],
    'NT': ['WA', 'SA', 'Q'],
    'SA': ['WA', 'NT', 'Q', 'NSW', 'V'],
    'Q':  ['NT', 'SA', 'NSW'],
    'NSW':['Q', 'SA', 'V'],
    'V':  ['SA', 'NSW'],
    'T':  []
}

colors = ['R', 'G', 'B']
```

```python
# Degree heuristic: pick node with most neighbors
def select_var(assignment):
    unassigned = [v for v in graph if v not in assignment]
    return max(unassigned, key=lambda v: len(graph[v]))


# Check constraint
def valid(var, color, assignment):
    return all(assignment.get(n) != color for n in graph[var])
```

```python
# Backtracking
def solve(assignment={}):
    if len(assignment) == len(graph):
        return assignment

    var = select_var(assignment)

    for color in colors:
        if valid(var, color, assignment):
            assignment[var] = color
            result = solve(assignment)
            if result:
                return result
            del assignment[var]

    return None


# Run
solution = solve()

print(solution)
```

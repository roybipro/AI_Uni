```python
# Simple Hill Climbing (Minimization, First Improvement, Alphabetical Order)

def hill_climbing(start, graph, heuristic):
    current = start
    path = [current]

    while True:
        # Generate successors in alphabetical order
        neighbors = sorted(graph.get(current, []))
        moved = False

        for node in neighbors:
            # Move to the first better neighbor (lower heuristic)
            if heuristic[node] < heuristic[current]:
                current = node
                path.append(current)
                moved = True
                break

        # Stop if no better neighbor found
        if not moved:
            break

    return path, current, heuristic[current]


# 🔹 Heuristic values
h = {
    'S': 100,
    'A': 95, 'B': 90, 'C': 105,
    'F': 96, 'G': 84,
    'H': 80,
    'D': 110, 'E': 85,
    'I': 109, 'J': 128
}

# 🔹 Graph (successor list)
graph = {
    'S': ['A', 'B', 'C'],
    'A': ['F', 'G'],
    'G': ['H'],
    'C': ['D', 'E'],
    'D': ['I', 'J']
}

# 🔹 Run Hill Climbing
path, final_node, cost = hill_climbing('S', graph, h)

# 🔹 Output
print("Expanded sequence:", " → ".join(path))
print("Final node:", final_node)
print("Heuristic value:", cost)
```

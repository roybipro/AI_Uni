```python
graph = {
    'A': ['B', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['B', 'D', 'F'],
    'D': ['A', 'C'],
    'E': ['B', 'F'],
    'F': ['C', 'E'],
}

colors = ['R', 'G', 'B']


# Check valid color
def valid(node, color, assignment):
    for neighbor in graph[node]:
        if assignment.get(neighbor) == color:
            return False
    return True


assignment = {}

# Degree heuristic ordering
nodes = sorted(graph, key=lambda x: len(graph[x]), reverse=True)

# Greedy coloring
for node in nodes:
    for color in colors:
        if valid(node, color, assignment):
            assignment[node] = color
            break

print(assignment)
```
